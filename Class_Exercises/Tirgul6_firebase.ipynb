```python
from firebase import firebase

FBconn = firebase.FirebaseApplication(
    'https://dean-cloud-db-default-rtdb.firebaseio.com/',
    None
)

print("Connected!")
```

```python
data = {"Temp": 25}
FBconn.post('/test/', data)
```

```python
res = FBconn.get('/test/', None)
print(res)
```

```python
def add_word(word):
    # מנסה לקרוא מהמיקום /words/
    words = FBconn.get('/words/', None) or {}

    # אם המילה כבר קיימת
    if word in words:
        new_count = words[word] + 1
    else:
        new_count = 1

    # מעדכן במסד
    FBconn.put('/words/', word, new_count)
    print(f"Added '{word}' → count = {new_count}")
```

```python
def show_words():
    words = FBconn.get('/words/', None)

    if not words:
        print("No words in DB yet.")
        return

    print("\n--- Words in DB ---")
    for word, count in words.items():
        print(f"{word}: {count}")
```

```python
def update_word(word, new_count):
    FBconn.put('/words/', word, new_count)
    print(f"Updated '{word}' → {new_count}")
```

```python
def delete_word(word):
    FBconn.delete('/words/', word)
    print(f"Deleted '{word}' from DB.")
```

```python
add_word("hello")
add_word("world")
add_word("hello")
show_words()
```

```python
while True:
    print("\n--- Words Menu ---")
    print("1. Add a single word")
    print("2. Add full text (auto count words)")
    print("3. Update word count manually")
    print("4. Delete a word")
    print("5. Show all words")
    print("6. Exit")

    choice = input("\nEnter your choice (1-6): ")

    # 1. Add one word
    if choice == "1":
        w = input("Enter a word: ").strip().lower()
        add_word(w)

    # 2. Add full text
    elif choice == "2":
        text = input("Enter full text: ").lower()
        words = text.split()
        for w in words:
            add_word(w)

    # 3. Update word count
    elif choice == "3":
        w = input("Enter the word to update: ").strip().lower()
        new = int(input("Enter new count: "))
        update_word(w, new)

    # 4. Delete a word
    elif choice == "4":
        w = input("Enter word to delete: ").strip().lower()
        delete_word(w)

    # 5. Show all
    elif choice == "5":
        show_words()

    # 6. Exit
    elif choice == "6":
        print("Goodbye!")
        break

    else:
        print("Invalid choice. Try again.")
```
